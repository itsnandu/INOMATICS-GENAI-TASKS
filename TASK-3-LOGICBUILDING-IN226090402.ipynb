```python
#Problem Statement 1: Employee Performance Bonus Eligibility
employees = {}

# number of employees
n = int(input("Enter number of employees: "))

# taking user input
for i in range(n):
    name = input("Enter employee name: ")
    score = int(input("Enter performance score: "))
    employees[name] = score

# Step 1: Find highest score
highest_score = max(employees.values())

# Step 2: Find employees with highest score
top_performers = [name for name, score in employees.items() if score == highest_score]

# Step 3: Display result
print("Top Performers Eligible for Bonus:", ", ".join(top_performers), f"(Score: {highest_score})")
```

```python
# Problem Statement 2: Search Query Keyword Analysis
import string

# user input
query = input("Enter search query: ")

# Step 1: convert to lowercase
query = query.lower()

# Step 2: remove punctuation
query = query.translate(str.maketrans('', '', string.punctuation))

# Step 3: split into words
words = query.split()

# Step 4: count frequency
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

# Step 5: display only keywords searched more than once
result = {word: count for word, count in freq.items() if count > 1}

print(result)
```

```python
#Problem Statement 3: Sensor Data Validation
# number of readings
n = int(input("Enter number of sensor readings: "))

sensor_readings = []

# taking readings from user
for i in range(n):
    value = int(input(f"Enter reading for hour {i}: "))
    sensor_readings.append(value)

# identify valid (even) readings
valid_readings = [(index, value) for index, value in enumerate(sensor_readings) if value % 2 == 0]

# display result
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
#Problem Statement 4: Email Domain Usage Analysis
emails = []

n = int(input("Enter number of emails: "))

# taking emails from user
for i in range(n):
    email = input("Enter email: ")
    emails.append(email)

# Step 1: count domains
domain_count = {}

for email in emails:
    domain = email.split("@")[1]
    domain_count[domain] = domain_count.get(domain, 0) + 1

# Step 2: calculate percentage
total = len(emails)

for domain, count in domain_count.items():
    percentage = (count / total) * 100
    print(f"{domain}: {percentage:.0f}%")
```

```python
#Problem Statement 5: Sales Spike Detection
sales = []

n = int(input("Enter number of days: "))

# taking sales input
for i in range(n):
    value = int(input(f"Enter sales for day {i+1}: "))
    sales.append(value)

# Step 1: calculate average sales
average_sales = sum(sales) / len(sales)

# Step 2: detect spikes (30% above average)
threshold = average_sales * 1.30

# Step 3: display spike days
for i, value in enumerate(sales):
    if value > threshold:
        print(f"Day {i+1}: {value}")
```
