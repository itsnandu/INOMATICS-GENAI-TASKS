```python
#Problem 1: Social Media – Post Engagement Analyzer
def analyze_post_engagement():
    total_likes = 0
    
    n = int(input("Enter number of like entries: "))
    
    # Loop to take user input
    for i in range(n):
        like = int(input(f"Enter likes {i+1}: "))
        total_likes += like
    
    # Check post status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    
    # Display output
    print("Total Likes:", total_likes)
    print("Post Status:", status)


# Function Call
analyze_post_engagement()
```

```python
#Problem 3: Agriculture – Rainfall Adequacy Checker
def rainfall_checker():
    
    n = int(input("Enter number of rainfall readings: "))
    total = 0
    
    for i in range(n):
        rain = float(input(f"Enter rainfall {i+1}: "))
        total += rain
    
    required_level = float(input("Enter required rainfall level: "))
    
    average = total / n
    
    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    
    print("Average Rainfall:", int(average))
    print("Rainfall Status:", status)


# Function Call
rainfall_checker()
```

```python
#Problem 4: Social Media – Duplicate Account Detection
def detect_duplicate_accounts():
    
    n = int(input("Enter number of usernames: "))
    usernames = []
    
    for i in range(n):
        name = input(f"Enter username {i+1}: ")
        usernames.append(name)
    
    if len(usernames) != len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Function Call
detect_duplicate_accounts()
```

```python
#Problem 6: Agriculture – Premium Crop Price Filter
def premium_crop_filter():
    
    n = int(input("Enter number of crop prices: "))
    prices = []
    
    for i in range(n):
        price = int(input(f"Enter price {i+1}: "))
        prices.append(price)
    
    premium_crops = []
    
    for price in prices:
        if price > 2000:
            premium_crops.append(price)
    
    print("Premium Crops:", premium_crops)


# Function Call
premium_crop_filter()
```

```python
#Problem 7: System Monitoring – Application Health Checker
def application_health_checker():
    
    error_count = int(input("Enter error count: "))
    
    if error_count == 0:
        status = "Healthy"
    elif error_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    
    print("Error Count:", error_count)
    print("System Status:", status)


# Function Call
application_health_checker()
```

```python
#Problem 8: Banking – Daily Transaction Limit Checker
def transaction_limit_checker():
    
    daily_limit = 50000
    amount = int(input("Enter transaction amount: "))
    
    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"
    
    print("Transaction Amount:", amount)
    print("Transaction Status:", status)


# Function Call
transaction_limit_checker()
```

```python
#Problem 9: E-Learning – Student Attendance Eligibility System
def attendance_eligibility():
    
    n = int(input("Enter total number of classes: "))
    attendance_list = []
    
    for i in range(n):
        status = input(f"Enter attendance for class {i+1} (P/A): ")
        attendance_list.append(status)
    
    attended_classes = 0
    
    for status in attendance_list:
        if status == "P":
            attended_classes += 1
    
    percentage = (attended_classes / n) * 100
    
    if percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"
    
    print("Attendance Percentage:", round(percentage, 1))
    print("Exam Eligibility:", eligibility)


# Function Call
attendance_eligibility()
```

```python
#Problem 10: Smart Electricity Bill Analyzer
def electricity_bill_analyzer():
    
    units = int(input("Enter units consumed: "))
    
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"
    
    print("Units Consumed:", units)
    print("Total Bill: ₹", bill)
    print("Usage Classification:", status)


electricity_bill_analyzer()
```
