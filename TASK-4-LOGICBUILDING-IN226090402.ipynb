```python
#Problem 1: Mobile Recharge Validation System
def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]
    
    if amount < 50:
        print("Minimum recharge amount is ₹50.")
        return False
    
    if amount not in valid_plans:
        print("Invalid plan selected.")
        print("Available plans: 199, 299, 399, 599")
        return False
    
    return True


# Retry system using while loop
while True:
    try:
        amount = int(input("Enter recharge amount: ₹"))
        
        if validate_recharge(amount):
            print("Recharge Successful!")
            break
        else:
            print("Please try again.\n")
    
    except ValueError:
        print("Please enter a valid number.\n")
```

```python
#Problem 2: Inventory Reorder Alert System
def check_inventory(stock_data):
    for product, quantity in stock_data.items():
        if quantity < 15:
            print(f"{product} → Reorder Alert (Stock: {quantity})")
        else:
            print(f"{product} → Stock OK (Stock: {quantity})")


# Sample inventory data
inventory = {
    "Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 18
}

check_inventory(inventory)
```

```python
#Problem 3: Student Result Processing System
class Student:
    
    def __init__(self, marks):
        self.marks = marks
    
    def calculate_average(self):
        total = 0
        
        # Loop to calculate total
        for mark in self.marks:
            total += mark
        
        return total / len(self.marks)
    
    def get_result(self):
        average = self.calculate_average()
        
        if average >= 50:
            return f"Average: {average:.2f} → Pass"
        else:
            return f"Average: {average:.2f} → Fail"


# Example usage
student1 = Student([60, 45, 70, 55])
print(student1.get_result())
```

```python
#Problem 4: Cab Fare Estimator with Retry Option
def calculate_fare(distance, peak_hour):
    base_fare = 50
    per_km_rate = 12
    
    fare = base_fare + (distance * per_km_rate)
    
    if peak_hour.lower() == "yes":
        fare += fare * 0.25   # 25% extra
    
    return fare


# Retry system using while loop
while True:
    try:
        distance = float(input("Enter distance in km: "))
        peak = input("Is it peak hour? (yes/no): ")
        
        total_fare = calculate_fare(distance, peak)
        print(f"Total Fare: ₹{total_fare:.2f}")
        
        retry = input("Do you want to calculate again? (yes/no): ")
        if retry.lower() != "yes":
            print("Thank you for using Cab Fare Estimator!")
            break
    
    except ValueError:
        print("Please enter valid input.\n")
```

```python
#Problem 5: Employee Attendance Eligibility Checker
def check_attendance(attendance_list):
    present_days = 0
    
    # Loop to count present days
    for day in attendance_list:
        if day.upper() == "P":
            present_days += 1
    
    total_days = len(attendance_list)
    percentage = (present_days / total_days) * 100
    
    if percentage >= 75:
        return f"Attendance: {percentage:.2f}% → Eligible"
    else:
        return f"Attendance: {percentage:.2f}% → Not Eligible"


# Taking user input
attendance_input = input("Enter attendance (P for Present, A for Absent) separated by space: ")

# Convert input string to list
attendance_list = attendance_input.split()

result = check_attendance(attendance_list)
print(result)
```

```python
#Problem 6: Password Strength Checker
def check_password_strength(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"
    
    # Check minimum length
    if len(password) < 8:
        return "Weak Password (Minimum 8 characters required)"
    
    # Loop over string
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True
    
    # Conditions
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password (Must contain at least 1 digit and 1 special character)"


# User input
password = input("Enter your password: ")
result = check_password_strength(password)
print(result)
```
