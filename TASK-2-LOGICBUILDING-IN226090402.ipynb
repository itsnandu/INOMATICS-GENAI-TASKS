```python
#Problem 1: Unique Words in a Sentence

# Input sentence
sentence = input("Enter sentence: ")

# Split sentence into words
words = sentence.split()

# Convert list to set to get unique words
unique_words = set(words)

# Output
print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
```

```python
#Problem 2: Highest Salary from Employee Data

# number of employees
n = int(input("Enter number of employees: "))

employees = {}

# taking user input
for i in range(n):
    name = input("Enter employee name: ")
    salary = float(input("Enter salary: "))
    employees[name] = salary

# finding highest salary manually
highest_employee = None
highest_salary = 0

for name, salary in employees.items():
    if salary > highest_salary:
        highest_salary = salary
        highest_employee = name

# output
print(f"Highest Salary: {highest_employee} - {highest_salary}")
```

```python
#Problem 3: Find Maximum and Minimum Values
# Take list input from user
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

# Initialize max and min with first element
maximum = numbers[0]
minimum = numbers[0]

# Traverse the list
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

# Output
print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

```python
#Problem 4: Count Products Above a Price Threshold
prices = list(map(int, input("Enter product prices separated by space: ").split()))

count = 0

for price in prices:
    if price > 1000:
        count += 1

print("Products above 1000:", count)
```

```python
#Problem 5: Calculate Attendance Percentage

attendance = input("Enter attendance (P/A) separated by space: ").split()

present_count = 0
total_days = len(attendance)

for day in attendance:
    if day.upper() == "P":
        present_count += 1

percentage = (present_count / total_days) * 100

print("Attendance Percentage:", percentage)
```

```python
#Problem 6: Remove Duplicate Phone Numbers

# Take input from user
phone_numbers = list(map(int, input("Enter phone numbers separated by space: ").split()))

# Convert to set to remove duplicates
unique_numbers = set(phone_numbers)

print("Unique phone numbers:", unique_numbers)
```

```python
#Problem 7: Count Character Frequency

# Input string
text = input("Enter a string: ")

char_count = {}

# Count frequency of each character
for ch in text:
    if ch in char_count:
        char_count[ch] += 1
    else:
        char_count[ch] = 1

# Output
print(char_count)
```

```python
#Problem 8: Convert List to Tuple

numbers = list(map(int, input("Enter numbers separated by space: ").split()))

numbers_tuple = tuple(numbers)

print("Tuple:", numbers_tuple)
```

```python
#Problem 9: Check if a Key Exists in a Dictionary

# Take dictionary input
n = int(input("Enter number of employees: "))
employees = {}

for i in range(n):
    name = input("Enter employee name: ")
    salary = int(input("Enter salary: "))
    employees[name] = salary

# Key to check
key = input("Enter employee name to search: ")

# Check existence
if key in employees:
    print("Employee exists")
else:
    print("Employee does not exist")
```

```python
#Problem 10: Calculate Average Marks

marks = list(map(float, input("Enter marks separated by space: ").split()))

total = 0
for mark in marks:
    total += mark

average = total / len(marks)

print("Average Marks:", average)
```
