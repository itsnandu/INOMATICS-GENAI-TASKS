```python
#Problem Statement 1:Smart Parking Lot Management System
def smart_parking(capacity, logs):
    current_vehicles = 0
    peak_usage = 0
    
    for log in logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            if current_vehicles > 0:
                current_vehicles -= 1
        
        # Track peak usage
        if current_vehicles > peak_usage:
            peak_usage = current_vehicles
    
    print("Currently Parked Vehicles:", current_vehicles)
    print("Peak Parking Usage:", peak_usage)
    
    # Check capacity
    if current_vehicles > capacity:
        print("Parking Status: Exceeded Capacity 🚨")
    else:
        print("Parking Status: Available ✅")


# Sample Input
capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking(capacity, vehicle_logs)
```

```python
#Problem Statement 2:Online Food Delivery Time Estimator
def estimate_delivery_time(distance, traffic, weather):
    # Base speed assumption: 30 km/hr
    base_speed = 30  
    
    # Calculate base time in minutes
    base_time = (distance / base_speed) * 60  
    
    # Traffic delay
    if traffic == "Low":
        base_time += 5
    elif traffic == "Medium":
        base_time += 10
    elif traffic == "High":
        base_time += 20
    
    # Weather delay
    if weather == "Clear":
        base_time += 0
    elif weather == "Rainy":
        base_time += 15
    elif weather == "Storm":
        base_time += 25
    
    print("Estimated Delivery Time:", int(base_time), "minutes")


# Sample Input
distance = 8
traffic = "High"
weather = "Rainy"

estimate_delivery_time(distance, traffic, weather)
```

```python
#Problem Statement 3:Movie Theatre Seat Occupancy Analyzer
def analyze_occupancy(total_seats, booked_seats):
    
    # Count booked seats
    booked_count = len(booked_seats)
    
    # Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100
    
    print("Occupancy:", int(occupancy), "%")
    
    # Determine show status
    if occupancy == 100:
        print("Show Status: Housefull 🎬")
        print("Suggestion: Open Additional Show")
        
    elif occupancy >= 75:
        print("Show Status: Almost Full 🔥")
        print("Suggestion: Consider Opening Extra Show")
        
    else:
        print("Show Status: Seats Available ✅")


# Sample Input
total_seats = 200
booked_seats = [1] * 150   # 150 booked seats

analyze_occupancy(total_seats, booked_seats)
```

```python
#Problem Statement 4:Cloud Server Load Classification System
def classify_server_load(cpu_readings):
    
    total = 0
    
    # Calculate total using loop
    for reading in cpu_readings:
        total += reading
    
    # Calculate average
    average = total / len(cpu_readings)
    
    print("Average CPU Load:", int(average), "%")
    
    # Classification
    if average < 50:
        print("Server Status: Normal ✅")
    elif 50 <= average < 80:
        print("Server Status: Warning ⚠️")
    else:
        print("Server Status: Critical 🚨")


# Sample Input
cpu_readings = [45, 60, 70, 85, 90]

classify_server_load(cpu_readings)
```

```python
#Problem Statement 5:Smart Classroom Resource Usage Monitor
def monitor_resources(resource_usage):
    
    overused = []
    threshold = 8   # Overuse limit (hours)
    
    # Loop through dictionary
    for resource, hours in resource_usage.items():
        if hours > threshold:
            overused.append(resource)
    
    # Output
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes ⚠️")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No ✅")


# Sample Input
usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_resources(usage)
```
