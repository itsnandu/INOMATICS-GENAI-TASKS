```python
#Problem 1: Smart Light Controller
class SmartLight:
    
    def __init__(self, name):
        self.name = name
        self.status = "OFF"   # Default state
    
    def turn_on(self):
        self.status = "ON"
        print(f"{self.name} is {self.status}")
    
    def turn_off(self):
        self.status = "OFF"
        print(f"{self.name} is {self.status}")
    
    def display_status(self):
        print(f"{self.name} is {self.status}")


# Sample Input
light = SmartLight("Bedroom Light")

# Action
light.turn_on()
```

```python
#Problem 2: Employee ID Card System
class Employee:
    
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    def display_id_card(self):
        print("\nEmployee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)


# Taking User Input
name = input("Enter Employee Name: ")
emp_id = input("Enter Employee ID: ")
department = input("Enter Department: ")

# Create Object
employee = Employee(name, emp_id, department)

# Display ID Card
employee.display_id_card()
```

```python
#Problem 3: Mobile Contact Record
class Contact:
    
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    
    def display_contact(self):
        print("\nContact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone)


# User Input
name = input("Enter Contact Name: ")
phone = input("Enter Phone Number: ")

# Create Object
contact1 = Contact(name, phone)

# Display Contact Info
contact1.display_contact()
```

```python
#Problem 5: Movie Rating Display System
class Movie:
    
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    
    def display_movie(self):
        print("\nMovie:", self.name)
        print("Rating:", self.rating, "/ 5")


# User Input
movie_name = input("Enter Movie Name: ")
rating = float(input("Enter Rating (out of 5): "))

# Create Object
movie1 = Movie(movie_name, rating)

# Display Details
movie1.display_movie()
```

```python
#Problem 6: Delivery Address Manager
class Delivery:
    
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address
    
    def print_details(self):
        print("\nDelivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.address)


# User Input
name = input("Enter Customer Name: ")
address = input("Enter Address: ")

# Create Object
delivery1 = Delivery(name, address)

# Print Delivery Details
delivery1.print_details()
```
